```python
import matplotlib.pyplot as plt 
import serial # Connect to ST-Link Virtual COM Port (replace 'COM3' with your port, e.g., '/dev/ttyACM0' on Mac/Linux) 
ser = serial.Serial('/dev/tty.usbmodem103', 115200, timeout=2) 
print("Listening for 200 samples on /dev/tty.usbmodem103...")
samples = [] 
while len(samples) < 200: 
    line = ser.readline().decode('utf-8').strip() 
    if line.isdigit(): 
        samples.append(int(line)) 
ser.close() # Plot 
print(f"Done! Collected {len(samples)} samples.")
    
plt.figure(figsize=(10, 4)) 
plt.plot(samples, 'o-')
plt.title('Live ADC Sample Buffer') 
plt.xlabel('Sample Index') 
plt.ylabel('ADC Raw Count') 
plt.grid(True) 
plt.show()
```
